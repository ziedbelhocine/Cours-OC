```python
# import des librairies

import plotly.offline as py 

import plotly.graph_objs as go

import pandas as pd 

import numpy as np

from plotly.figure_factory import create_table 

import plotly.express as px
```

```python
# importation des données

gapminder = px.data.gapminder()
```

```python
type(gapminder)
```

```python
gapminder
```

```python
# diagramme à barres

df_canada = gapminder[gapminder['country'] == 'Canada']
df_canada
```

```python
fig = px.bar(df_canada, x='year', y='pop', title='Evolution de la population du Canada',
             hover_data=['lifeExp', 'gdpPercap'],
             color='lifeExp')
fig.show()
```

```python
# scatterplot
# espérance de vie vs gdp

gapminder_2007 = gapminder.query("year == 2007")

gapminder_2007
```

```python
fig = px.scatter(gapminder_2007, y='lifeExp', x='gdpPercap', title='Espérance de vie vs gdp',
             hover_data=['pop', 'country'],
             color='continent')
fig.show()
```

```python
# animation

fig = px.scatter(gapminder, y='lifeExp', x='gdpPercap', 
        title='Espérance de vie vs gdp',
        hover_data=['pop', 'country'],
        size='pop',
        facet_col='continent',
        color='continent',
        animation_frame='year',
        animation_group='country')

fig.show()
```
