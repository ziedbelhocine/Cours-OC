```python
#Importation de la librairie Pandas
import pandas as pd
#Importation de Seaborn
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Importation du fichier web.xlsx
df_gender = pd.read_csv("marts_gender.csv")
#Importation du fichier erp.xlsx
df_age = pd.read_csv("marts_age.csv")
#Importation du fichier liaison.xlsx
df_region = pd.read_csv("marts_region.csv")
```

```python
df_age.head(5)
```

```python
sns.lineplot(data=df_age, x="YEAR", y="DELTA_PCT", hue="AGE_GROUP")
plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.)
plt.title("Evolution de la repartition des ages chez les étudiants en Data d'OPC comparé à leur repartition dans la population française")
plt.grid(axis='y')
plt.xticks(rotation=45, ha='right')
plt.ylabel('Delta en %')
plt.xlabel('Année')
plt.show()
```

```python
sns.lineplot(data=df_gender, x="YEAR", y="PCT", hue="GENDER")
plt.legend(bbox_to_anchor=(1.05, 1), loc=2, borderaxespad=0.)
plt.title("Evolution de la répartition des genres déclarés chez les étudiants en Data d'OPC et ceux de la population française")
plt.grid(axis='y')
plt.xticks(rotation=45, ha='right')
plt.ylabel('%')
plt.xlabel('Année')
plt.show()
```
