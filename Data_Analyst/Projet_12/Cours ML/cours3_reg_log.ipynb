```python
from sklearn.datasets import load_breast_cancer
X, y = load_breast_cancer(return_X_y=True)

from sklearn.linear_model import LogisticRegression
clf = LogisticRegression(random_state=808, max_iter=10000).fit(X, y)
```

'X' est un array (n lignes et p colonnes) avec les valeurs des variables prédictrices et 'y' un array(n lignes, 1 colonne) des de la variable cible pour chaque individu.

```python
X.shape
```

```python
y.shape
```

```python
print(clf.predict([X[8, :]]))
print(y[8])
```

```python
clf.predict_proba([X[8, :]])
```

```python
y_hat_proba = clf.predict_proba(X)[:,1]
import seaborn as sns
sns.histplot(y_hat_proba)
```

```python
clf.predict_proba(X)
```

```python
y_pred = clf.predict(X)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_pred, y)
```

Matrics de confusion donne TN, FP / FN, TP

```python
from sklearn.metrics import confusion_matrix
confusion_matrix(y, y_pred)
```

On test nos métriques d'éval avec des t = 0.3 ou 0.7 ou 0.5

```python
y_pred_03 = [ 0 if value < 0.3 else 1 for value in y_hat_proba ]
y_pred_07 = [ 0 if value < 0.7 else 1 for value in y_hat_proba ]
```

```python
confusion_matrix(y, y_pred_03)
```

```python
confusion_matrix(y, y_pred_07)
```

```python
from sklearn.metrics import precision_score, recall_score
print(accuracy_score(y, y_pred))
print(precision_score(y, y_pred))
print(recall_score(y, y_pred))
```

```python
print(accuracy_score(y, y_pred_03))
print(precision_score(y, y_pred_03))
print(recall_score(y, y_pred_03))
```

```python
print(accuracy_score(y, y_pred_07))
print(precision_score(y, y_pred_07))
print(recall_score(y, y_pred_07))
```

```python
from sklearn.metrics import roc_curve
fpr, tpr, thresholds = roc_curve(y, y_hat_proba)

import matplotlib.pyplot as plt
plt.plot(fpr, tpr)
```
