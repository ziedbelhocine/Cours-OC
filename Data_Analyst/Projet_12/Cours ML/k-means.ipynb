```python
from sklearn.datasets import make_blobs
centers = [[2, 2], [-2, -2], [2, -2]]
X, labels_true = make_blobs(
                      n_samples=3000, 
                      centers=centers, 
                      cluster_std=0.7)
```

X est un array à 2 dimensions avec 3000 individus, répartis en 3 clusters dont les centroides sont les coordonnées dans centers. L'ecart-type au sein de chaque cluster est de 0.7.
Labels,_true est une array a 1 dimension contenant les clusters auxquels appartiennent les individus de X.

```python
from sklearn.cluster import KMeans
k_means = KMeans( n_clusters=3, random_state = 808)
k_means.fit(X)
print(k_means.cluster_centers_)
```

si on lance un k-means sur X. On retrouve pour centre des clusters les coordonnées de nos centres initiaux.

```python
k_means.score(X)
```
