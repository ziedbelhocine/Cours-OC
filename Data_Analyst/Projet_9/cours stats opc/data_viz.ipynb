```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Chargement des données
df_clients = pd.read_csv('Données+clients+VertiGo+nettoyées.csv')
df_ets = pd.read_csv('Données+entreprise+VertiGo+nettoyées.csv')
```

```python
plt.figure(figsize=(10, 6))
plt.hist(df_clients['prix total'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution du Prix Total des Voyages')
plt.xlabel('Prix Total')
plt.ylabel('Nombre de Voyages')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
voyages_par_type = df_clients['type de voyage'].value_counts()

voyages_par_type
```

```python
# Création du graphique à barres
plt.figure(figsize=(10, 6))
voyages_par_type.plot(kind='bar', color='teal', edgecolor='black')
plt.title('Nombre de Voyages par Type de Voyage')
plt.xlabel('Type de Voyage')
plt.ylabel('Nombre de Voyages')
plt.xticks(rotation=0)  
# Rotation des étiquettes pour une meilleure lisibilité
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
df_clients['date de début de voyage'] = pd.to_datetime(df_clients['date de début de voyage'], dayfirst=True)  # Assure-toi que le format de la date est correct
```

```python
# Trier les données par date de début de voyage
data_sorted = df_clients.sort_values('date de début de voyage')
```

```python
import matplotlib.dates as mdates

# Création du graphique en ligne
plt.figure(figsize=(12, 7))
plt.plot(data_sorted['date de début de voyage'], data_sorted['prix total'], marker="o", linestyle='-', color='royalblue', label="Prix")
plt.plot(data_sorted['date de début de voyage'], data_sorted['prix total'].rolling(30).mean(), linestyle='-', color='red', label="Moyenne mobile 30 jours")
# Formater l'axe des X pour afficher les dates correctement
plt.gca().xaxis.set_major_formatter(mdates.DateFormatter('%d-%m-%Y'))
plt.gca().xaxis.set_major_locator(mdates.AutoDateLocator())
plt.gcf().autofmt_xdate()  # Rotation automatique des dates pour une meilleure lisibilité
plt.legend()

plt.title('Évolution du Prix Total des Voyages en Fonction de la Date de Début')
plt.xlabel('Date de Début de Voyage')
plt.ylabel('Prix Total')
plt.grid(True)
plt.show()
```

```python
# Créer un nuage de points
plt.figure(figsize=(10, 6))
plt.scatter(df_clients['durée de voyage (en jours)'], df_clients['prix total'], color='darkorange', edgecolor='black', alpha=0.7)

plt.title('Relation entre la Durée du Voyage et le Prix Total')
plt.xlabel('Durée du Voyage (en jours)')
plt.ylabel('Prix Total')
plt.grid(True, which='both', linestyle='--', linewidth=0.5)
plt.show()
```

```python
age_median = df_clients['âge'].median()
print(f"L'âge médian des Voyageur(ses) est de {age_median}")

plt.figure(figsize=(10, 6))
plt.hist(df_clients['âge'], bins=20, color='skyblue', edgecolor='black')
plt.scatter(age_median, 0, color='darkorange', edgecolor='black', s=100, marker='x', label ='Médiane')
plt.title('Distribution de l\'âge des Voyageur(ses)')
plt.xlabel('âge')
plt.ylabel('Nombre de clients')
plt.legend()
plt.grid(axis='y', alpha=0.75)
plt.show()
```
