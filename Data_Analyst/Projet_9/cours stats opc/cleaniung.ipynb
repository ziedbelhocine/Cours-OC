```python
col_quantit = ['âge', 'durée de voyage (en jours)', 'nombre de personnes', 'prix total', 'évaluation sur 5', 'budget_voyage_annuel', 'durée moyenne de voyage annuelle (en jours)']
# création fonction detect outliers 

def detect_outliers(df, feature):
    q1 = df[feature].quantile(0.25)
    q3 = df[feature].quantile(0.75)
    iqr = q3-q1
    out_bas = (q1 -2*iqr)
    out_haut = (q3 + 2*iqr)

    return [((df[feature] < out_bas) | (df[feature] > out_haut)), iqr, out_bas, out_haut]

for i in col_quantit:
    print(f"--- Outliers pour la colonne : {i} ---")
    outliers = df_clients[detect_outliers(df_clients, i)[0]]
    display(outliers)
```

```python
df_clients.loc[df_clients['âge'] == 350, 'âge'] = 35
df_clients.loc[df_clients['âge'] == 220, 'âge'] = 22
df_clients.loc[df_clients['âge'] == 4100, 'âge'] = 41
```

```python
values = {}

for c in df_clients.columns:
    if df_clients[c].dtype != 'object':
        values[c] = df_clients[c].median()
    else :
        values[c] = df_clients[c].mode()[0]

df_clients.fillna(value=values, inplace=True)
```
