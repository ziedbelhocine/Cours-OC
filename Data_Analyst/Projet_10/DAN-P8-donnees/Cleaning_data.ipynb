```python
# On complète les trous par les valeurs passées ou futures du MÊME pays et de la MÊME granularité
df_services = df_services.sort_values(by=['Country', 'Granularity', 'Year'])

df_services['Population using at least basic drinking-water services (%)'] = (
    df_services
    .groupby(['Country', 'Granularity'])['Population using at least basic drinking-water services (%)']
    .transform(lambda x: x.ffill().bfill())
)
```

```python
# 1. Dictionnaire de secours pour les orthographes "pièges" ou non-standards
custom_mapping = {
    'Democratic Republic of the Congo': 'COD',
    'Republic of the Congo': 'COG',
    'Congo': 'COG',
    'Congo, Democratic Republic of the': 'COD',
    'Congo (Dem. Rep.)': 'COD',
    'Congo (Rep.)': 'COG',
    'Bolivia (Plurinational State of)': 'BOL',
    'Venezuela (Bolivarian Republic of)': 'VEN',
    'Iran (Islamic Republic of)': 'IRN',
    'Viet Nam': 'VNM',
    'Republic of Korea': 'KOR',
    'Micronesia (Federated States of)': 'FSM',
    'Turkey': 'TUR'
}

def get_iso3(country_name):
    # On nettoie les espaces en trop
    country_name = str(country_name).strip()
    
    # Étape A : On vérifie si le pays est dans notre dictionnaire de secours
    if country_name in custom_mapping:
        return custom_mapping[country_name]
    
    # Étape B : Recherche automatique dans la base ISO officielle
    try:
        # Recherche exacte ou par nom commun
        country = pycountry.countries.lookup(country_name)
        return country.alpha_3 # Retourne le code ISO à 3 lettres (ex: FRA, USA)
        # Note : remplace par .alpha_2 si tu préfères le code à 2 lettres (ex: FR, US)
    except LookupError:
        # Si pycountry ne trouve pas, on renvoie None pour repérer le trou
        return None

# 3. Application directe sur ton df_region pour créer la nouvelle colonne
df_region['ISO_Code'] = df_region['Country'].apply(get_iso3)

# 4. Vérification des pays qui n'ont pas trouvé de correspondance
missing_countries = df_region[df_region['ISO_Code'].isna()]['Country'].unique()
print("Pays non reconnus :", missing_countries)
```
