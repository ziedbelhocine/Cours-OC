```python
import pandas as pd
import numpy as np
```

```python
#Importation des fichiers
df_services = pd.read_csv("BasicAndSafelyManagedDrinkingWaterServices.csv")
df_mortality = pd.read_csv("MortalityRateAttributedToWater.csv")
df_politic = pd.read_csv("PoliticalStability.csv")
df_population = pd.read_csv("Population.csv")
df_region = pd.read_csv("RegionCountry.csv")
```

```python
df_region.rename(columns={'REGION (DISPLAY)': 'Region', 'COUNTRY (DISPLAY)' : 'Country'}, inplace=True)
```

```python
df_population.loc[df_population['Country'] == 'North Macedonia'] = 'Republic of North Macedonia'
df_politic.loc[df_politic['Country'] == 'North Macedonia'] = 'Republic of North Macedonia'
```

```python
df_services.head()
manquants_services = list(set(df_population['Country'] ) - set(df_services['Country']))
manquants_population = list(set(df_services['Country'] ) - set(df_population['Country']))
print(f"product_id dans df_population mais pas dans df_services: {len(manquants_services)}")
print(f"product_id dans df_services mais pas dans df_population: {len(manquants_population)}")
```
